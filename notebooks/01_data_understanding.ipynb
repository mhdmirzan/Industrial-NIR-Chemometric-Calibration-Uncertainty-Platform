### Detect wavelength columns

```python
def numeric_column_value(col):
    try:
        return float(str(col).strip())
    except:
        return np.nan


column_wavelengths = {
    col: numeric_column_value(col)
    for col in df.columns
}

wavelength_candidates = {
    col: wavelength
    for col, wavelength in column_wavelengths.items()
    if not np.isnan(wavelength)
}
```

```python
wavelength_values = np.array(
    list(wavelength_candidates.values())
)

print(
    "Minimum candidate wavelength:",
    wavelength_values.min()
)

print(
    "Maximum candidate wavelength:",
    wavelength_values.max()
)
```

```python
sorted(
    wavelength_candidates.items(),
    key = lambda x: x[1]
)[:20]
```

```python
spectral_columns = [
    col
    for col, wavelength in wavelength_candidates.items()
    if 900 <= wavelength <= 1700
]
```

### Extract wavelengths

```python
wavelengths = np.array([
    float(str(col).strip())
    for col in spectral_columns
])

sort_idx = np.argsort(wavelengths)

wavelengths = wavelengths[sort_idx]

spectral_columns = [
    spectral_columns[i]
    for i in sort_idx
]
```

```python
print("Number of wavelengths:", len(wavelengths))
print("First wavelength:", wavelengths[0])
print("Last wavelength:", wavelengths[-1])
```

### Inspect wavelength spacing

```python
spacing = np.diff(wavelengths)

print("Minimum spacing: ", spacing.min())
print("Maximum spacing: ", spacing.max())
```

Therefore the spacing gap is consistent which is 5

### Spectral mean and standard deviation

```python
spectral_mean = np.nanmean(X_spectral, axis=0)

spectral_std = np.nanstd(X_spectral, axis=0)
```

```python
spectral_mean
```

```python
# Mean plot

plt.figure(figsize=(14,6))

plt.plot(wavelengths, spectral_mean)

plt.xlabel("Wavelength (nm)")
plt.ylabel("Mean spectral values")
plt.title("Mean NIR Spectrum")

plt.show()
```

```python
# Variability plot

plt.figure(figsize=(14,6))

plt.plot(wavelengths, spectral_std)

plt.xlabel("Wavelength (nm)")
plt.ylabel("Standard deviation")
plt.title("Spectral Variability")

plt.show()
```

### Mean ± standard deviation

```python
plt.figure(figsize=(14,6))

plt.plot(wavelengths, spectral_mean, label='Mean')

plt.fill_between(
    wavelengths,
    spectral_mean - spectral_std,
    spectral_mean + spectral_std,
    alpha = 0.2,
    label = "±1 SD"
)

plt.xlabel("Wavelength (nm)")
plt.ylabel("Spectral response")
plt.title("Mean Spectrum ± 1 Standard Deviation")
plt.legend()

plt.show()
```

This explains where your dataset has high spectral variation.

### Spectral Outlier detection

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
X_temp = SimpleImputer(strategy='median').fit_transform(X_spectral)
```

```python
X_scaled = StandardScaler().fit_transform(X_temp)
```

```python
pca = PCA(n_components=5)

X_pca = pca.fit_transform(X_scaled)
```

```python
print(pca.explained_variance_ratio_)

print("Total: ", pca.explained_variance_ratio_.sum())
```

Since, the first 2 components combined and result to 99.39%. these are the important features. but if we consider all five components, the total importance comes upto 99.96%

```python
plt.figure(figsize=(9,7))

plt.scatter(
    X_pca[:,0],
    X_pca[:,1],
    alpha=0.7
)

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA Score Plot")

plt.show()
```

### PCA colored by crop

```python
plt.figure(figsize=(10,7))

for crop in df[CROP_COL].dropna().unique():

    mask = (df[CROP_COL]== crop)

    plt.scatter(
        X_pca[mask,0],
        X_pca[mask,1],
        label = str(crop),
        alpha=0.6
    )

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA Score Plot by Crop")
plt.legend()

plt.show()
```
